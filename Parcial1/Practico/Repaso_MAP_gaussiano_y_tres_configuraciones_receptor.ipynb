# Repaso de parcial: MAP gaussiano y tres configuraciones del receptor

Este notebook resume dos temas que pueden aparecer como pregunta teórica en hoja escrita:

1. Demostrar la regla MAP para dos hipótesis gaussianas, incluso cuando las varianzas son distintas.
2. Demostrar las tres configuraciones equivalentes del receptor MAP para canal AWGN de formas de onda.

Referencia conceptual: apunte de Cabrera, Unidad 2 y Unidad 3, especialmente diseño del receptor, regla MAP, arquitectura transmisor-receptor y estructuras receptoras alternativas.

## 1. Por qué MAP minimiza la probabilidad de error

La regla MAP significa **maximum a posteriori**, es decir, elegir la hipótesis más probable después de observar el dato recibido.

Si el receptor observa $Y=y$, y decide $\hat H=i$, entonces la probabilidad de acertar con esa decisión es:

$$
P(H=i\mid Y=y).
$$

Para ese valor fijo de $y$, el receptor no puede cambiar la observación; solamente puede elegir qué hipótesis declarar. Entonces, para maximizar la probabilidad de acierto punto a punto, debe elegir la hipótesis con mayor probabilidad posterior:

$$
\boxed{\hat H_{MAP}(y)=\arg\max_i P(H=i\mid Y=y)}
$$

Como maximizar la probabilidad de acierto equivale a minimizar la probabilidad de error, MAP es la regla óptima cuando todos los errores tienen el mismo costo.

## 2. Demostración MAP para dos gaussianas con varianzas distintas

Supongamos dos hipótesis:

$$
H=0: \quad Y\sim\mathcal N(c_0,\sigma_0^2)
$$

$$
H=1: \quad Y\sim\mathcal N(c_1,\sigma_1^2)
$$

con probabilidades a priori:

$$
P_H(0)=p_0,\qquad P_H(1)=p_1.
$$

La regla MAP decide $H=1$ si:

$$
P(H=1\mid Y=y)\ge P(H=0\mid Y=y).
$$

Aplicando Bayes:

$$
\frac{f_{Y\mid H}(y\mid 1)P_H(1)}{f_Y(y)}
\ge
\frac{f_{Y\mid H}(y\mid 0)P_H(0)}{f_Y(y)}.
$$

Como $f_Y(y)>0$, se cancela en ambos lados:

$$
f_{Y\mid H}(y\mid 1)p_1
\ge
f_{Y\mid H}(y\mid 0)p_0.
$$

Esta es la forma práctica de MAP: **verosimilitud multiplicada por prior**.

## 3. Sustitución de las densidades gaussianas

Para cada hipótesis:

$$
f_{Y\mid H}(y\mid i)=\frac{1}{\sqrt{2\pi}\sigma_i}
\exp\left[-\frac{(y-c_i)^2}{2\sigma_i^2}\right].
$$

Entonces MAP decide $H=1$ si:

$$
\frac{p_1}{\sqrt{2\pi}\sigma_1}
\exp\left[-\frac{(y-c_1)^2}{2\sigma_1^2}\right]
\ge
\frac{p_0}{\sqrt{2\pi}\sigma_0}
\exp\left[-\frac{(y-c_0)^2}{2\sigma_0^2}\right].
$$

Paso 1. Cancelamos $\sqrt{2\pi}$:

$$
\frac{p_1}{\sigma_1}
\exp\left[-\frac{(y-c_1)^2}{2\sigma_1^2}\right]
\ge
\frac{p_0}{\sigma_0}
\exp\left[-\frac{(y-c_0)^2}{2\sigma_0^2}\right].
$$

Paso 2. Aplicamos logaritmo natural, que no cambia el sentido porque es creciente:

$$
\ln p_1-\ln\sigma_1-\frac{(y-c_1)^2}{2\sigma_1^2}
\ge
\ln p_0-\ln\sigma_0-\frac{(y-c_0)^2}{2\sigma_0^2}.
$$

Paso 3. Reordenamos dejando los términos cuadráticos del lado izquierdo:

$$
\frac{(y-c_0)^2}{2\sigma_0^2}
-
\frac{(y-c_1)^2}{2\sigma_1^2}
\ge
\ln\left(\frac{p_0\sigma_1}{p_1\sigma_0}\right).
$$

Por lo tanto:

$$
\boxed{
\hat H=1
\quad\text{si}\quad
\frac{(y-c_0)^2}{2\sigma_0^2}
-
\frac{(y-c_1)^2}{2\sigma_1^2}
\ge
\ln\left(\frac{p_0\sigma_1}{p_1\sigma_0}\right)
}
$$

y se decide $\hat H=0$ en caso contrario.

## 4. Forma cuadrática para resolver en hoja

Para resolver a mano conviene llevar todo a una función $g(y)$:

$$
g(y)=
\frac{(y-c_0)^2}{2\sigma_0^2}
-
\frac{(y-c_1)^2}{2\sigma_1^2}
-
\ln\left(\frac{p_0\sigma_1}{p_1\sigma_0}\right).
$$

La regla queda:

$$
\boxed{\hat H=1 \quad\text{si}\quad g(y)\ge0}
$$

$$
\boxed{\hat H=0 \quad\text{si}\quad g(y)<0}
$$

Al expandir:

$$
g(y)=Ay^2+By+C
$$

donde:

$$
A=\frac{1}{2\sigma_0^2}-\frac{1}{2\sigma_1^2}
$$

$$
B=-\frac{c_0}{\sigma_0^2}+\frac{c_1}{\sigma_1^2}
$$

$$
C=\frac{c_0^2}{2\sigma_0^2}-\frac{c_1^2}{2\sigma_1^2}
-\ln\left(\frac{p_0\sigma_1}{p_1\sigma_0}\right).
$$

Detalle importante: si $\sigma_0^2\ne\sigma_1^2$, la frontera puede no ser un único umbral. Puede aparecer una región de decisión entre dos raíces o fuera de dos raíces, porque la desigualdad es cuadrática.

## 5. Caso especial: varianzas iguales

Si $\sigma_0^2=\sigma_1^2=\sigma^2$, desaparece el término cuadrático. La regla vuelve a ser de un solo umbral.

Partimos de:

$$
\frac{(y-c_0)^2}{2\sigma^2}
-
\frac{(y-c_1)^2}{2\sigma^2}
\ge
\ln\left(\frac{p_0}{p_1}\right).
$$

Sacamos factor común:

$$
\frac{(y-c_0)^2-(y-c_1)^2}{2\sigma^2}
\ge
\ln\left(\frac{p_0}{p_1}\right).
$$

Expandimos la diferencia de cuadrados:

$$
(y-c_0)^2-(y-c_1)^2
=2y(c_1-c_0)+c_0^2-c_1^2.
$$

También puede escribirse como:

$$
2y(c_1-c_0)+c_0^2-c_1^2
=(c_1-c_0)(2y-c_0-c_1).
$$

Entonces:

$$
\frac{(c_1-c_0)(2y-c_0-c_1)}{2\sigma^2}
\ge
\ln\left(\frac{p_0}{p_1}\right).
$$

Si $c_1>c_0$, dividimos sin cambiar el sentido:

$$
y\ge
\frac{c_0+c_1}{2}
+
\frac{\sigma^2}{c_1-c_0}
\ln\left(\frac{p_0}{p_1}\right).
$$

Por lo tanto:

$$
\boxed{
\gamma_{MAP}=
\frac{c_0+c_1}{2}
+
\frac{\sigma^2}{c_1-c_0}
\ln\left(\frac{p_0}{p_1}\right)
}
$$

Si además $p_0=p_1$, entonces $\ln(p_0/p_1)=0$ y queda el umbral ML clásico:

$$
\boxed{\gamma_{ML}=\frac{c_0+c_1}{2}}.
$$

## 6. Tres configuraciones equivalentes del receptor MAP

Ahora pasamos al caso de formas de onda en AWGN continuo.

Cuando $H=j$, se transmite la forma de onda $w_j(t)$ y el receptor observa:

$$
R(t)=w_j(t)+N(t).
$$

Se elige una base ortonormal $\{\psi_1(t),\ldots,\psi_n(t)\}$ para representar las señales. El receptor forma la estadística suficiente:

$$
Y_k=\langle R,\psi_k\rangle,
\qquad k=1,\ldots,n.
$$

Así se obtiene el vector:

$$
Y=(Y_1,\ldots,Y_n)^T.
$$

Cada forma de onda $w_j(t)$ queda representada por un vector código $c_j$:

$$
c_j=(c_{j,1},\ldots,c_{j,n})^T.
$$

El problema continuo se transforma en un problema discreto vectorial:

$$
H=j: \quad Y=c_j+Z,
$$

donde:

$$
Z\sim\mathcal N\left(0,\frac{N_0}{2}I_n\right).
$$

A partir de ahí salen las tres configuraciones equivalentes.

## 7. Configuración 1: receptor por distancia euclídea corregida por prior

La densidad condicional es:

$$
f_{Y\mid H}(y\mid j)=
\frac{1}{(\pi N_0)^{n/2}}
\exp\left[-\frac{\lVert y-c_j\rVert^2}{N_0}\right].
$$

MAP elige el $j$ que maximiza:

$$
P_H(j)f_{Y\mid H}(y\mid j).
$$

Tomamos logaritmo:

$$
\ln P_H(j)-\frac{\lVert y-c_j\rVert^2}{N_0}
$$

El término $-(n/2)\ln(\pi N_0)$ no aparece porque no depende de $j$.

Maximizar esa expresión equivale a minimizar su versión multiplicada por $-N_0$:

$$
\boxed{
\hat H_{MAP}(y)=
\arg\min_j
\left[\lVert y-c_j\rVert^2-N_0\ln P_H(j)\right]
}
$$

Esta es la primera configuración: el receptor forma $Y$ y luego elige la menor distancia corregida por el prior.

## 8. Configuración 2: receptor por producto interno con los vectores código

Partimos de la métrica anterior:

$$
\lVert y-c_j\rVert^2-N_0\ln P_H(j).
$$

Expandimos la distancia:

$$
\lVert y-c_j\rVert^2
=
\lVert y\rVert^2-2\langle y,c_j\rangle+\lVert c_j\rVert^2.
$$

Sustituyendo:

$$
\lVert y\rVert^2-2\langle y,c_j\rangle+\lVert c_j\rVert^2-N_0\ln P_H(j).
$$

Como $\lVert y\rVert^2$ no depende de $j$, se elimina para comparar hipótesis.

Minimizar:

$$
-2\langle y,c_j\rangle+\lVert c_j\rVert^2-N_0\ln P_H(j)
$$

equivale a maximizar, multiplicando por $-1/2$:

$$
\boxed{
\hat H_{MAP}(y)=
\arg\max_j
\left[
\langle y,c_j\rangle
-\frac{\lVert c_j\rVert^2}{2}
+\frac{N_0}{2}\ln P_H(j)
\right]
}
$$

Esta es la segunda configuración: el receptor forma $Y$ y luego calcula productos internos contra cada vector código $c_j$.

## 9. Configuración 3: receptor con correladores o filtros adaptados a las formas de onda

Cada forma de onda puede escribirse en la base ortonormal como:

$$
w_j(t)=\sum_{k=1}^n c_{j,k}\psi_k(t).
$$

Recordemos que:

$$
Y_k=\langle R,\psi_k\rangle.
$$

Entonces el producto interno vectorial puede escribirse como:

$$
\langle y,c_j\rangle
=
\sum_{k=1}^n y_kc_{j,k}.
$$

Sustituimos $y_k=\langle R,\psi_k\rangle$:

$$
\langle y,c_j\rangle
=
\sum_{k=1}^n c_{j,k}\langle R,\psi_k\rangle.
$$

Por linealidad del producto interno:

$$
\sum_{k=1}^n c_{j,k}\langle R,\psi_k\rangle
=
\left\langle R,\sum_{k=1}^n c_{j,k}\psi_k\right\rangle.
$$

Pero la suma es justamente $w_j(t)$:

$$
\langle y,c_j\rangle=\langle R,w_j\rangle.
$$

En forma integral:

$$
\langle R,w_j\rangle=\int r(t)w_j^*(t)\,dt.
$$

Además, por Parseval:

$$
\lVert c_j\rVert^2=\lVert w_j\rVert^2.
$$

Entonces la segunda configuración se transforma en:

$$
\boxed{
\hat H_{MAP}(r)=
\arg\max_j
\left[
\int r(t)w_j^*(t)\,dt
-\frac{\lVert w_j\rVert^2}{2}
+\frac{N_0}{2}\ln P_H(j)
\right]
}
$$

Esta es la tercera configuración: el receptor compara correlaciones directas entre la señal recibida y cada forma de onda posible. Esa integral se puede implementar con correladores o con filtros adaptados.

## 10. Correlador y filtro adaptado

Las tres configuraciones necesitan operaciones del tipo:

$$
\int r(t)b^*(t)\,dt.
$$

Hay dos formas típicas de implementarlas.

**Correlador:** multiplica $r(t)$ por $b^*(t)$ e integra.

$$
\int r(t)b^*(t)\,dt
$$

**Filtro adaptado:** usa un filtro con respuesta al impulso:

$$
h(t)=b^*(T-t).
$$

La salida del filtro es:

$$
y(t)=\int r(\alpha)b^*(T+\alpha-t)\,d\alpha.
$$

Muestreando en $t=T$:

$$
y(T)=\int r(\alpha)b^*(\alpha)\,d\alpha.
$$

Es decir, el filtro adaptado entrega la misma cantidad que el correlador, pero como salida de un sistema lineal filtrante.

## 11. Resumen para escribir en el parcial

Si te piden MAP gaussiano escalar con varianzas distintas:

$$
\hat H=1
\quad\text{si}\quad
\frac{(y-c_0)^2}{2\sigma_0^2}
-
\frac{(y-c_1)^2}{2\sigma_1^2}
\ge
\ln\left(\frac{p_0\sigma_1}{p_1\sigma_0}\right).
$$

Si las varianzas son iguales y $c_1>c_0$:

$$
\hat H=1
\quad\text{si}\quad
y\ge
\frac{c_0+c_1}{2}
+
\frac{\sigma^2}{c_1-c_0}\ln\left(\frac{p_0}{p_1}\right).
$$

Si te piden las tres configuraciones del receptor MAP en AWGN:

1. Formar $Y$ y minimizar $\lVert y-c_j\rVert^2-N_0\ln P_H(j)$.
2. Formar $Y$ y maximizar $\langle y,c_j\rangle-\lVert c_j\rVert^2/2+(N_0/2)\ln P_H(j)$.
3. Trabajar directo con formas de onda y maximizar $\int r(t)w_j^*(t)\,dt-\lVert w_j\rVert^2/2+(N_0/2)\ln P_H(j)$.

La demostración clave es:

$$
\lVert y-c_j\rVert^2=\lVert y\rVert^2-2\langle y,c_j\rangle+\lVert c_j\rVert^2
$$

y:

$$
\langle y,c_j\rangle=\langle R,w_j\rangle=\int r(t)w_j^*(t)\,dt.
$$
